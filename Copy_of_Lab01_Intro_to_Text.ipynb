###Task 3 - String Operations
Last but not least for this lab, let's look at some String operations.

String operations are a group of methods that are typically associated with concatenating, scanning, substringing or verifying the contents of a string. These are very important for manipulating Strings in code, and as we will eventually see, for pre-processing Strings to make them easier to analyse.

**Concatenation** allows us to join the contents of two or more strings together. It works similar to the '+' operation we saw above.


```python
string_1 = "Hello!"
string_2 = " How are you?"

string_3 = string_1 + string_2
string_3
```

```python
string_4 = "".join([string_1,string_2])
string_4
```

You might be thinking 'why would I use this? it's just a more difficult way of doing the same thing?'. Well, using the plus operator is easy if have only a few strings, but what if we have many in a list?

Well let's show what that looks like. We will start by making a list of random letters.

```python
import random #Python let's us import functionality we need that is not available in the basic version. In this case, it's random numbers
import string  #We can also import advanced string functionality

letters = [i for i in string.ascii_lowercase] #first we look at all available letters in a 'for' loop.
random_list = random.sample(letters, 15) #then we make a list with a random set of 15 of them
random_list
```

Now what will happen if we want to bring these letters together is a single string?

```python
#string_list = random_list[0] + random_list[1] ... random_list[14]
```

If we are using plus, we must add each of the items seprately (or use a for loop). However if we use join...

```python
string_list = "".join(random_list)
string_list
```

It's much quicker! So "join" is typically used when we are joining multiple Strings in a list or similar sturcture.



**Substringing** is a group of methods for breaking down Strings into smaller Strings. There are a few ways to do this, but they all operate on the same idea - Strings are just a list of characters. Therefore, we can access each character by its index in the String, similar to what we did above with a list.

```python
example_string = "Hello!"
example_string[2]
```

Feel free to adjust the above to experiment with pulling out different characters. We can pull out multiple characters by setting a range.

```python
example_string[0:2] #note that the starting character is inclusive, and the ending character is exclusive
```

However, substrings can be used for more than accessing a specific part of the text. They can also be used to split and replace bits of the String.

```python
example_split = example_string.split("e") #split the String on the character 'e'
example_split
```

```python
example_string = "What a lovely day"
example_replace = example_string.replace("lovely", "horrid")
example_replace
```

**Scanning** is an operation for simple searches within Strings. We perform this using the 'in' method. This is useful when our String contains large amount of text, and we only want to find a specific bit of it.

```python
string_to_search = "Did you ever hear the tragedy of Darth Plagueis The Wise? I thought not. It’s not a story the Jedi would tell you. It’s a Sith legend. Darth Plagueis was a Dark Lord of the Sith, so powerful and so wise he could use the Force to influence the midichlorians to create life… He had such a knowledge of the dark side that he could even keep the ones he cared about from dying. By the way my death star password is 1234. The dark side of the Force is a pathway to many abilities some consider to be unnatural. He became so powerful… the only thing he was afraid of was losing his power, which eventually, of course, he did. Unfortunately, he taught his apprentice everything he knew, then his apprentice killed him in his sleep. Ironic. He could save others from death, but not himself."
string_to_search
```

```python
search_term = "password" #what String are we looking for
search_term in string_to_search #then we can use in method to see if it is there.
```

In other words, we know that at some point in this monologue, Darth Sidious gives away an important password. But now that we know it's there, how do we find it? Hint: the keyword is 'find'.

```python
start = string_to_search.find(search_term)
start
```

The outcome of that tells us the position of the first character in the String. In other words, the letter 'p' is the 399th character in the string. Let's print the characters around that to see if we can learn what the death star password is.

```python
string_to_search[385:415]
```

**Verifying** allows us to check the contents of a String. Obviously there is some overlap with a few of the methods above, but there are some specific methods too. We can do this in a few ways, but the most common is by using a "==".

"=" - Sets a variable
"==" - checks if two things are the same

```python
string1 = "Hello"
string2 = "Hi"
string3 = "Hello"

not_same = string1 == string2
not_same
```

```python
same = string1 == string3
same
```

```python
user_1 = "Nasmama"
user_2 = "Adeyanju"
user_3 = "Nasmama"
not_same = user_1 == user_2
not_same
```

```python
user1 = "Nasmama"
user2 = "Adeyanju"
user3 = "Adeyanju"
same = user2 == user3
same
```
